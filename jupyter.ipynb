```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

RANDOM_STATE = 20240504


# Define the path
data_path = Path("data/hotel_bookings.csv")


# Load the CSV file
hotel = pd.read_csv(data_path)


hotel.head()

# Display the dimensions of the dataset
print(f"Rows: {hotel.shape[0]:,}")
print(f"Columns: {hotel.shape[1]}")


hotel.info()


n_duplicates = hotel.duplicated().sum()
duplicate_pct = hotel.duplicated().mean()

print(f"Exact duplicate rows: {n_duplicates:,}")  # number of duplicate rows
print(f"Percentage duplicated: {duplicate_pct:.2%}") # proportion of duplicate rows

# Remove exact duplicate observations while keeping the first occurrence
hotel_clean = hotel.drop_duplicates().copy()

# Verify the dimensions and remaining duplicate count
print(f"Original observations: {len(hotel):,}")
print(f"Observations after removing duplicates: {len(hotel_clean):,}")
print(f"Observations removed: {len(hotel) - len(hotel_clean):,}")
print(f"Remaining exact duplicates: {hotel_clean.duplicated().sum():,}")

# Create an initial data-quality summary for each variable
audit = pd.DataFrame({
    "dtype": hotel_clean.dtypes.astype(str),
    "missing_n": hotel_clean.isna().sum(),
    "missing_pct": hotel_clean.isna().mean() * 100,
    "unique_n": hotel_clean.nunique()
})


# Show in descending order
audit = audit.sort_values("missing_pct", ascending=False)

audit

# company: convert to binary indicator, drop raw ID (94% missing, too sparse to use directly)
hotel_clean["has_company"] = hotel_clean["company"].notna().astype(int)
hotel_clean = hotel_clean.drop(columns=["company"])

print(f"'has_company' created; original 'company' column dropped.")

# agent: encode missing values as "No Agent" and treat IDs as categories ( unique IDs will be grouped later)
hotel_clean["agent"] = (
    hotel_clean["agent"]
    .astype("Int64")
    .astype("string")
    .fillna("No Agent")
)


# country: encode missing values as "Unknown Country" (keeps unknown countries separate from rare observed countries)
hotel_clean["country"] = (
    hotel_clean["country"]
    .fillna("Unknown Country")
)


print(hotel_clean[["agent", "country"]].isna().sum())





#Examine the relationship between reservation status and cancellation
status_table = pd.crosstab(
    hotel_clean["reservation_status"],
    hotel_clean["is_canceled"],
    margins=True
)

display(status_table)

# Calculate cancellation proportions within each reservation status
status_proportions = pd.crosstab(
    hotel_clean["reservation_status"],
    hotel_clean["is_canceled"],
    normalize="index"
)

display(status_proportions)


# Remove variables containing post-outcome information
leakage_variables = [
    "reservation_status",
    "reservation_status_date"
]

hotel_clean = hotel_clean.drop(columns=leakage_variables)

print(f"Remaining variables: {hotel_clean.shape[1]}")

# Summarize numerical variables to inspect ranges and extreme values
numeric_summary = hotel_clean.describe().T

display(numeric_summary)

# Summarize potentially suspicious or unusual observations
suspicious_summary = pd.DataFrame({
    "Check": [
        "Zero adults",
        "Zero total guests",
        "Zero total nights",
        "Negative ADR",
        "Adults above 20",
        "Children above 3",
        "Babies above 2"
    ],
    "Count": [
        (hotel_clean["adults"] == 0).sum(),

        (
            (hotel_clean["adults"] == 0) &
            (hotel_clean["children"] == 0) &
            (hotel_clean["babies"] == 0)
        ).sum(),

        (
            hotel_clean["stays_in_weekend_nights"] +
            hotel_clean["stays_in_week_nights"] == 0
        ).sum(),

        (hotel_clean["adr"] < 0).sum(),
        (hotel_clean["adults"] > 20).sum(),
        (hotel_clean["children"] > 3).sum(),
        (hotel_clean["babies"] > 2).sum()
    ]
})

display(suspicious_summary)



# Examine bookings with zero adults
zero_adults = hotel_clean[
    hotel_clean["adults"] == 0
]

# Determine how many zero-adult bookings also contain no guests
zero_guests = zero_adults[
    (zero_adults["children"] == 0) &
    (zero_adults["babies"] == 0)
]

# Zero-adult bookings with at least one recorded child or baby
zero_adults_with_dependents = zero_adults[
    (zero_adults["children"] > 0) |
    (zero_adults["babies"] > 0)
]

print(f"Bookings with zero adults: {len(zero_adults):,}")
print(f"Confirmed zero-guest bookings: {len(zero_guests):,}")
print(
    f"Zero-adult bookings with children or babies: "
    f"{len(zero_adults_with_dependents):,}"
)



hotel_clean = hotel_clean.drop(index=zero_guests.index).copy()

print(f"Observations removed: {len(zero_guests):,}")

print(f"Observations remaining: {len(hotel_clean):,}")



# Identify bookings with zero recorded nights
zero_nights = hotel_clean[
    (   hotel_clean["stays_in_weekend_nights"] +
        hotel_clean["stays_in_week_nights"]
    ) == 0
]

print(f"Zero-night bookings: {len(zero_nights):,}")

# Examine a sample of zero-night bookings
display(
    zero_nights[
        [
            "hotel",
            "is_canceled",
            "lead_time",
            "adults",
            "children",
            "babies",
            "adr",
            "market_segment",
            "customer_type"
        ]
    ].head(20)
)


print(
    f"Zero-night bookings with ADR = 0: "
    f"{(zero_nights['adr'] == 0).sum():,}"
)




# Examine characteristics of zero-night bookings

zero_night_details = zero_nights.copy()

zero_night_details["reservation_status"] = hotel.loc[
    zero_night_details.index, "reservation_status"
]

print("Reservation status:")
display(
    zero_night_details["reservation_status"]
    .value_counts(normalize=True)
    .to_frame("proportion")
)


print("Market segment:")
display(
    zero_night_details["market_segment"]
    .value_counts(normalize=True)
    .to_frame("proportion")
)

print("Customer type:")
display(
    zero_night_details["customer_type"]
    .value_counts(normalize=True)
    .to_frame("proportion")
)

print("Distribution channel:")
display(
    zero_night_details["distribution_channel"]
    .value_counts(normalize=True)
    .to_frame("proportion")
)

# Flag bookings with potentially missing/miscoded stay information
hotel_clean["questionable_stay_data"] = (
    (
        hotel_clean["stays_in_weekend_nights"] +
        hotel_clean["stays_in_week_nights"]
    ) == 0
) & (hotel_clean["adr"] == 0)

# Create adjusted versions for analyses involving ADR or length of stay
hotel_clean["adr_adj"] = hotel_clean["adr"].where(
    ~hotel_clean["questionable_stay_data"],
    np.nan
)

hotel_clean["total_nights_adj"] = (
    hotel_clean["stays_in_weekend_nights"] +
    hotel_clean["stays_in_week_nights"]
).where(
    ~hotel_clean["questionable_stay_data"],
    np.nan
)

print(
    f"Bookings flagged for potentially missing stay data: "
    f"{hotel_clean['questionable_stay_data'].sum():,}"
)

# Identify bookings with a negative ADR
negative_adr = hotel_clean[
    hotel_clean["adr"] < 0
]

print(f"Bookings with negative ADR: {len(negative_adr):,}")

display(
    negative_adr[
        [
            "hotel",
            "is_canceled",
            "adults",
            "children",
            "babies",
            "stays_in_weekend_nights",
            "stays_in_week_nights",
            "adr",
            "market_segment",
            "customer_type",
            "deposit_type"
        ]
    ]
)

# Treat the negative ADR as missing in the adjusted variable
hotel_clean.loc[negative_adr.index, "adr_adj"] = np.nan

print(
    f"Negative values remaining in adr_adj: "
    f"{(hotel_clean['adr_adj'] < 0).sum():,}"
)

display(
    hotel_clean["adr_adj"]
    .quantile([0.50, 0.90, 0.95, 0.99, 0.999, 1.00])
    .to_frame("adr_adj")
)

display(
    hotel_clean.nlargest(10, "adr_adj")[
        [
            "hotel", "adults", "children", "babies",
            "stays_in_weekend_nights", "stays_in_week_nights",
            "adr_adj", "market_segment", "customer_type",
            "is_canceled"
        ]
    ]
)

# Identify the isolated high-ADR outlier
extreme_adr = hotel_clean[hotel_clean["adr_adj"] > 1000]

print(f"Bookings with adr_adj above 1,000: {len(extreme_adr):,}")

# Treat the outlier as missing in the adjusted variable
hotel_clean.loc[hotel_clean["adr_adj"] > 1000, "adr_adj"] = np.nan

print(
    f"ADR values above 1,000 remaining in adr_adj: "
    f"{(hotel_clean['adr_adj'] > 1000).sum():,}"
)

# Show bookings with the largest numbers of adults
hotel_clean.nlargest(10, "adults")[
    ["hotel", "adults", "children", "babies",
     "market_segment", "customer_type", "adr", "is_canceled"]
]




# Examine the distribution of the number of children
display(
    hotel_clean["children"]
    .value_counts(dropna=False)
    .sort_index(na_position="last")
    .to_frame("count")
)

# Examine bookings with unusually large numbers of children
display(
    hotel_clean.loc[
        hotel_clean["children"] > 3,
        [
            "hotel",
            "adults",
            "children",
            "babies",
            "market_segment",
            "customer_type",
            "adr",
            "is_canceled"
        ]
    ]
)

extreme_children = hotel_clean[
    hotel_clean["children"] > 3
]

print(f"Bookings with more than three children: {len(extreme_children):,}")

# Remove the isolated outlier in number of children
hotel_clean = hotel_clean.drop(index=extreme_children.index).copy()


print(f"Observations removed: {len(extreme_children):,}")
print(f"Observations remaining: {len(hotel_clean):,}")



# Examine the distribution of the number of babies
display(
    hotel_clean["babies"]
    .value_counts()
    .sort_index()
    .to_frame("count")
)


# Examine bookings with unusually large numbers of babies
extreme_babies = hotel_clean[
    hotel_clean["babies"] > 2
]

print(f"Bookings with more than two babies: {len(extreme_babies):,}")

display(
    extreme_babies[
        [
            "hotel",
            "adults",
            "children",
            "babies",
            "market_segment",
            "customer_type",
            "adr",
            "is_canceled"
        ]
    ]
)

# Remove bookings with implausibly large numbers of babies
hotel_clean = hotel_clean.drop(index=extreme_babies.index).copy()

print(f"Observations removed: {len(extreme_babies):,}")
print(f"Observations remaining: {len(hotel_clean):,}")

from sklearn.model_selection import train_test_split

X = hotel_clean.drop(columns=["is_canceled"])
y = hotel_clean["is_canceled"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=RANDOM_STATE
)

train_df = X_train.copy()
train_df["is_canceled"] = y_train

print(f"Training set: {len(train_df):,} observations")
print(f"Test set: {len(X_test):,} observations")


# Calculate cancellation proportions for plotting

target_summary = (
    train_df["is_canceled"]
    .value_counts()
    .sort_index()
    .to_frame("count")
)

target_summary["proportion"] = (
    target_summary["count"] / len(train_df)
)

display(target_summary)

# Plot the distribution of the response variable
plt.figure(figsize=(6, 4))
plt.bar(["Not Canceled", "Canceled"], target_summary["proportion"])

plt.ylabel("Proportion")
plt.xlabel("Booking Status")
plt.title("Distribution of Hotel Booking Cancellations")
plt.ylim(0, 1)

plt.show()





# Group lead time and calculate cancellation rates
lead_time_group = pd.cut(
    train_df["lead_time"],
    bins=[-1, 7, 30, 90, 180, 365, np.inf],
    labels=["0–7", "8–30", "31–90", "91–180", "181–365", "366+"]
)

lead_time_summary = (
    train_df.groupby(lead_time_group, observed=False)["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(lead_time_summary)



lead_time_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(8, 5),
    xlabel="Lead Time (Days)",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Lead Time",
    rot=0
)

plt.gca().yaxis.set_major_formatter(PercentFormatter(1))
plt.tight_layout()
plt.show()



# Summarize cancellations by deposit type
deposit_summary = (
    train_df.groupby("deposit_type")["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
    .sort_values("cancellation_rate")
)

display(deposit_summary)

# Plot cancellation rate by deposit type
deposit_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 5),
    xlabel="Deposit Type",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Deposit Type",
    rot=0
)

plt.gca().yaxis.set_major_formatter(PercentFormatter(1))
plt.ylim(0, 1)
plt.tight_layout()
plt.show()

# Group previous cancellations into interpretable categories
previous_cancellation_group = pd.cut(
    train_df["previous_cancellations"],
    bins=[-1, 0, 1, np.inf],
    labels=["0", "1", "2+"]
)

previous_cancellation_summary = (
    train_df.groupby(
        previous_cancellation_group,
        observed=False
    )["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(previous_cancellation_summary)

# Plot cancellation rate by previous cancellation history
previous_cancellation_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 5),
    xlabel="Number of Previous Cancellations",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Previous Cancellation History",
    rot=0
)

plt.gca().yaxis.set_major_formatter(PercentFormatter(1))
plt.ylim(0, 1)
plt.tight_layout()
plt.show()

# Group previous non-canceled bookings into interpretable categories
previous_booking_group = pd.cut(
    train_df["previous_bookings_not_canceled"],
    bins=[-1, 0, 1, 5, np.inf],
    labels=["0", "1", "2–5", "6+"]
)

previous_booking_summary = (
    train_df.groupby(
        previous_booking_group,
        observed=False
    )["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(previous_booking_summary)

# Plot cancellation rate by previous booking history
previous_booking_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 5),
    xlabel="Previous Non-Canceled Bookings",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Previous Booking History",
    rot=0
)

plt.gca().yaxis.set_major_formatter(PercentFormatter(1))
plt.ylim(0, 1)
plt.tight_layout()
plt.show()

# Group the number of special requests into interpretable categories
special_request_group = pd.cut(
    train_df["total_of_special_requests"],
    bins=[-1, 0, 1, 2, np.inf],
    labels=["0", "1", "2", "3+"]
)

special_request_summary = (
    train_df.groupby(
        special_request_group,
        observed=False
    )["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(special_request_summary)

# Plot cancellation rate by number of special requests
special_request_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 5),
    xlabel="Number of Special Requests",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Number of Special Requests",
    rot=0
)


for i, rate in enumerate(special_request_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.2%}",
        ha="center"
    )


plt.gca().yaxis.set_major_formatter(PercentFormatter(1))
plt.ylim(0, 0.4)
plt.tight_layout()
plt.show()


# Group booking changes into interpretable categories
booking_change_group = pd.cut(
    train_df["booking_changes"],
    bins=[-1, 0, 1, 2, np.inf],
    labels=["0", "1", "2", "3+"]
)

booking_change_summary = (
    train_df.groupby(
        booking_change_group,
        observed=False
    )["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(booking_change_summary)

# Plot cancellation rate by number of booking changes
ax = booking_change_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 5),
    xlabel="Number of Booking Changes",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Number of Booking Changes",
    rot=0
)

ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 0.5)

for i, rate in enumerate(booking_change_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.2%}",
        ha="center"
    )

plt.tight_layout()
plt.show()




# Summarize cancellation rates by market segment
market_segment_summary = (
    train_df.groupby("market_segment")["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
    .sort_values("cancellation_rate")
)

display(market_segment_summary)


# Exclude extremely small categories from the visualization only
market_segment_plot = market_segment_summary[
    market_segment_summary["bookings"] >= 100
]


# Plot cancellation rate by market segment
ax = market_segment_plot["cancellation_rate"].plot(
    kind="barh",
    figsize=(8, 5),
    xlabel="Cancellation Rate",
    ylabel="Market Segment",
    title="Cancellation Rate by Market Segment (n>=100)"
)


ax.xaxis.set_major_formatter(PercentFormatter(1))
ax.set_xlim(0, 0.4)

for i, rate in enumerate(market_segment_plot["cancellation_rate"]):
    ax.text(rate + 0.005, i, f"{rate:.2%}", va="center")


plt.tight_layout()
plt.show()



# Audit the original parking-space categories
parking_raw_audit = (
    train_df.groupby("required_car_parking_spaces")["is_canceled"]
    .agg(
        bookings="size",
        cancellations="sum",
        cancellation_rate="mean"
    )
)

display(parking_raw_audit)

# More than 99% of positive values are 1, so pool sparse
# positive categories into a single binary indicator
has_parking_request = train_df["required_car_parking_spaces"] > 0

parking_summary = (
    train_df.groupby(has_parking_request)["is_canceled"]
    .agg(bookings="size", cancellation_rate="mean")
)

display(parking_summary)

# Plot cancellation rate by parking-space requirement
ax = parking_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(6, 4),
    xlabel="Requires Parking Space",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Parking-Space Requirement",
    rot=0
)
ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 0.4)

for i, rate in enumerate(parking_summary["cancellation_rate"]):
    ax.text(i, rate + 0.01, f"{rate:.2%}", ha="center")

plt.tight_layout()
plt.show()


# Check whether the complete-separation pattern holds across
# hotel type and arrival year
parking_hotel_audit = (
    train_df.assign(has_parking_request=has_parking_request)
    .groupby(["hotel", "has_parking_request"])["is_canceled"]
    .agg(bookings="size", cancellations="sum", cancellation_rate="mean")
)

parking_year_audit = (
    train_df.assign(has_parking_request=has_parking_request)
    .groupby(["arrival_date_year", "has_parking_request"])["is_canceled"]
    .agg(bookings="size", cancellations="sum", cancellation_rate="mean")
)

# Summarize the zero-cancellation pattern across hotels and years
parking_stability = pd.concat(
    {
        "Hotel": parking_hotel_audit.xs(True, level="has_parking_request"),
        "Arrival Year": parking_year_audit.xs(True, level="has_parking_request")
    },
    names=["dimension", "group"]
)

display(parking_stability)



# Compare adr_adj distributions by cancellation status
adr_summary = train_df.groupby("is_canceled")["adr_adj"].describe()
display(adr_summary)

# Visualize the distribution

ax = train_df.boxplot(
    column="adr_adj",
    by="is_canceled",
    figsize=(6, 4),
    showfliers=False # Hide fliers
)   


plt.suptitle("")
plt.title("ADR by Cancellation Status")
plt.xlabel("Is Canceled (0 = No, 1 = Yes)")
plt.ylabel("Average Daily Rate")
plt.tight_layout()
plt.show()


# Summarize cancellation rates by hotel type
hotel_summary = (
    train_df.groupby("hotel")["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(hotel_summary)

# Plot cancellation rate by hotel type
ax = hotel_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(6, 4),
    xlabel="Hotel Type",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Hotel Type",
    rot=0
)

ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(
    0,
    min(1, hotel_summary["cancellation_rate"].max() + 0.1)
)

for i, rate in enumerate(hotel_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.2%}",
        ha="center"
    )

plt.tight_layout()
plt.show()

# Summarize cancellation rates by customer type
customer_type_summary = (
    train_df.groupby("customer_type")["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
    .sort_values("cancellation_rate")
)

display(customer_type_summary)

# Plot cancellation rate by customer type
ax = customer_type_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 4),
    xlabel="Customer Type",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Customer Type",
    rot=0
)

ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(
    0,
    min(1, customer_type_summary["cancellation_rate"].max() + 0.1)
)

for i, rate in enumerate(customer_type_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.1%}",
        ha="center"
    )

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

RANDOM_STATE = 20240504


# Define the path
data_path = Path("data/hotel_bookings.csv")


# Load the CSV file
hotel = pd.read_csv(data_path)


hotel.head()

# Display the dimensions of the dataset
print(f"Rows: {hotel.shape[0]:,}")
print(f"Columns: {hotel.shape[1]}")


hotel.info()


n_duplicates = hotel.duplicated().sum()
duplicate_pct = hotel.duplicated().mean()

print(f"Exact duplicate rows: {n_duplicates:,}")  # number of duplicate rows
print(f"Percentage duplicated: {duplicate_pct:.2%}") # proportion of duplicate rows

# Remove exact duplicate observations while keeping the first occurrence
hotel_clean = hotel.drop_duplicates().copy()

# Verify the dimensions and remaining duplicate count
print(f"Original observations: {len(hotel):,}")
print(f"Observations after removing duplicates: {len(hotel_clean):,}")
print(f"Observations removed: {len(hotel) - len(hotel_clean):,}")
print(f"Remaining exact duplicates: {hotel_clean.duplicated().sum():,}")

# Create an initial data-quality summary for each variable
audit = pd.DataFrame({
    "dtype": hotel_clean.dtypes.astype(str),
    "missing_n": hotel_clean.isna().sum(),
    "missing_pct": hotel_clean.isna().mean() * 100,
    "unique_n": hotel_clean.nunique()
})


# Show in descending order
audit = audit.sort_values("missing_pct", ascending=False)

audit

# company: convert to binary indicator, drop raw ID (94% missing, too sparse to use directly)
hotel_clean["has_company"] = hotel_clean["company"].notna().astype(int)
hotel_clean = hotel_clean.drop(columns=["company"])

print(f"'has_company' created; original 'company' column dropped.")

# agent: encode missing values as "No Agent" and treat IDs as categories ( unique IDs will be grouped later)
hotel_clean["agent"] = (
    hotel_clean["agent"]
    .astype("Int64")
    .astype("string")
    .fillna("No Agent")
)


# country: encode missing values as "Unknown Country" (keeps unknown countries separate from rare observed countries)
hotel_clean["country"] = (
    hotel_clean["country"]
    .fillna("Unknown Country")
)


print(hotel_clean[["agent", "country"]].isna().sum())





#Examine the relationship between reservation status and cancellation
status_table = pd.crosstab(
    hotel_clean["reservation_status"],
    hotel_clean["is_canceled"],
    margins=True
)

display(status_table)

# Calculate cancellation proportions within each reservation status
status_proportions = pd.crosstab(
    hotel_clean["reservation_status"],
    hotel_clean["is_canceled"],
    normalize="index"
)

display(status_proportions)


# Remove variables containing post-outcome information
leakage_variables = [
    "reservation_status",
    "reservation_status_date"
]

hotel_clean = hotel_clean.drop(columns=leakage_variables)

print(f"Remaining variables: {hotel_clean.shape[1]}")

# Summarize numerical variables to inspect ranges and extreme values
numeric_summary = hotel_clean.describe().T

display(numeric_summary)

# Summarize potentially suspicious or unusual observations
suspicious_summary = pd.DataFrame({
    "Check": [
        "Zero adults",
        "Zero total guests",
        "Zero total nights",
        "Negative ADR",
        "Adults above 20",
        "Children above 3",
        "Babies above 2"
    ],
    "Count": [
        (hotel_clean["adults"] == 0).sum(),

        (
            (hotel_clean["adults"] == 0) &
            (hotel_clean["children"] == 0) &
            (hotel_clean["babies"] == 0)
        ).sum(),

        (
            hotel_clean["stays_in_weekend_nights"] +
            hotel_clean["stays_in_week_nights"] == 0
        ).sum(),

        (hotel_clean["adr"] < 0).sum(),
        (hotel_clean["adults"] > 20).sum(),
        (hotel_clean["children"] > 3).sum(),
        (hotel_clean["babies"] > 2).sum()
    ]
})

display(suspicious_summary)



# Examine bookings with zero adults
zero_adults = hotel_clean[
    hotel_clean["adults"] == 0
]

# Determine how many zero-adult bookings also contain no guests
zero_guests = zero_adults[
    (zero_adults["children"] == 0) &
    (zero_adults["babies"] == 0)
]

# Zero-adult bookings with at least one recorded child or baby
zero_adults_with_dependents = zero_adults[
    (zero_adults["children"] > 0) |
    (zero_adults["babies"] > 0)
]

print(f"Bookings with zero adults: {len(zero_adults):,}")
print(f"Confirmed zero-guest bookings: {len(zero_guests):,}")
print(
    f"Zero-adult bookings with children or babies: "
    f"{len(zero_adults_with_dependents):,}"
)



hotel_clean = hotel_clean.drop(index=zero_guests.index).copy()

print(f"Observations removed: {len(zero_guests):,}")

print(f"Observations remaining: {len(hotel_clean):,}")



# Identify bookings with zero recorded nights
zero_nights = hotel_clean[
    (   hotel_clean["stays_in_weekend_nights"] +
        hotel_clean["stays_in_week_nights"]
    ) == 0
]

print(f"Zero-night bookings: {len(zero_nights):,}")

# Examine a sample of zero-night bookings
display(
    zero_nights[
        [
            "hotel",
            "is_canceled",
            "lead_time",
            "adults",
            "children",
            "babies",
            "adr",
            "market_segment",
            "customer_type"
        ]
    ].head(20)
)


print(
    f"Zero-night bookings with ADR = 0: "
    f"{(zero_nights['adr'] == 0).sum():,}"
)




# Examine characteristics of zero-night bookings

zero_night_details = zero_nights.copy()

zero_night_details["reservation_status"] = hotel.loc[
    zero_night_details.index, "reservation_status"
]

print("Reservation status:")
display(
    zero_night_details["reservation_status"]
    .value_counts(normalize=True)
    .to_frame("proportion")
)


print("Market segment:")
display(
    zero_night_details["market_segment"]
    .value_counts(normalize=True)
    .to_frame("proportion")
)

print("Customer type:")
display(
    zero_night_details["customer_type"]
    .value_counts(normalize=True)
    .to_frame("proportion")
)

print("Distribution channel:")
display(
    zero_night_details["distribution_channel"]
    .value_counts(normalize=True)
    .to_frame("proportion")
)

# Flag bookings with potentially missing/miscoded stay information
hotel_clean["questionable_stay_data"] = (
    (
        hotel_clean["stays_in_weekend_nights"] +
        hotel_clean["stays_in_week_nights"]
    ) == 0
) & (hotel_clean["adr"] == 0)

# Create adjusted versions for analyses involving ADR or length of stay
hotel_clean["adr_adj"] = hotel_clean["adr"].where(
    ~hotel_clean["questionable_stay_data"],
    np.nan
)

hotel_clean["total_nights_adj"] = (
    hotel_clean["stays_in_weekend_nights"] +
    hotel_clean["stays_in_week_nights"]
).where(
    ~hotel_clean["questionable_stay_data"],
    np.nan
)

print(
    f"Bookings flagged for potentially missing stay data: "
    f"{hotel_clean['questionable_stay_data'].sum():,}"
)

# Identify bookings with a negative ADR
negative_adr = hotel_clean[
    hotel_clean["adr"] < 0
]

print(f"Bookings with negative ADR: {len(negative_adr):,}")

display(
    negative_adr[
        [
            "hotel",
            "is_canceled",
            "adults",
            "children",
            "babies",
            "stays_in_weekend_nights",
            "stays_in_week_nights",
            "adr",
            "market_segment",
            "customer_type",
            "deposit_type"
        ]
    ]
)

# Treat the negative ADR as missing in the adjusted variable
hotel_clean.loc[negative_adr.index, "adr_adj"] = np.nan

print(
    f"Negative values remaining in adr_adj: "
    f"{(hotel_clean['adr_adj'] < 0).sum():,}"
)

display(
    hotel_clean["adr_adj"]
    .quantile([0.50, 0.90, 0.95, 0.99, 0.999, 1.00])
    .to_frame("adr_adj")
)

display(
    hotel_clean.nlargest(10, "adr_adj")[
        [
            "hotel", "adults", "children", "babies",
            "stays_in_weekend_nights", "stays_in_week_nights",
            "adr_adj", "market_segment", "customer_type",
            "is_canceled"
        ]
    ]
)

# Identify the isolated high-ADR outlier
extreme_adr = hotel_clean[hotel_clean["adr_adj"] > 1000]

print(f"Bookings with adr_adj above 1,000: {len(extreme_adr):,}")

# Treat the outlier as missing in the adjusted variable
hotel_clean.loc[hotel_clean["adr_adj"] > 1000, "adr_adj"] = np.nan

print(
    f"ADR values above 1,000 remaining in adr_adj: "
    f"{(hotel_clean['adr_adj'] > 1000).sum():,}"
)

# Show bookings with the largest numbers of adults
hotel_clean.nlargest(10, "adults")[
    ["hotel", "adults", "children", "babies",
     "market_segment", "customer_type", "adr", "is_canceled"]
]




# Examine the distribution of the number of children
display(
    hotel_clean["children"]
    .value_counts(dropna=False)
    .sort_index(na_position="last")
    .to_frame("count")
)

# Examine bookings with unusually large numbers of children
display(
    hotel_clean.loc[
        hotel_clean["children"] > 3,
        [
            "hotel",
            "adults",
            "children",
            "babies",
            "market_segment",
            "customer_type",
            "adr",
            "is_canceled"
        ]
    ]
)

extreme_children = hotel_clean[
    hotel_clean["children"] > 3
]

print(f"Bookings with more than three children: {len(extreme_children):,}")

# Remove the isolated outlier in number of children
hotel_clean = hotel_clean.drop(index=extreme_children.index).copy()


print(f"Observations removed: {len(extreme_children):,}")
print(f"Observations remaining: {len(hotel_clean):,}")



# Examine the distribution of the number of babies
display(
    hotel_clean["babies"]
    .value_counts()
    .sort_index()
    .to_frame("count")
)


# Examine bookings with unusually large numbers of babies
extreme_babies = hotel_clean[
    hotel_clean["babies"] > 2
]

print(f"Bookings with more than two babies: {len(extreme_babies):,}")

display(
    extreme_babies[
        [
            "hotel",
            "adults",
            "children",
            "babies",
            "market_segment",
            "customer_type",
            "adr",
            "is_canceled"
        ]
    ]
)

# Remove bookings with implausibly large numbers of babies
hotel_clean = hotel_clean.drop(index=extreme_babies.index).copy()

print(f"Observations removed: {len(extreme_babies):,}")
print(f"Observations remaining: {len(hotel_clean):,}")

from sklearn.model_selection import train_test_split

X = hotel_clean.drop(columns=["is_canceled"])
y = hotel_clean["is_canceled"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=RANDOM_STATE
)

train_df = X_train.copy()
train_df["is_canceled"] = y_train

print(f"Training set: {len(train_df):,} observations")
print(f"Test set: {len(X_test):,} observations")


# Calculate cancellation proportions for plotting

target_summary = (
    train_df["is_canceled"]
    .value_counts()
    .sort_index()
    .to_frame("count")
)

target_summary["proportion"] = (
    target_summary["count"] / len(train_df)
)

display(target_summary)


# Plot the distribution of the response variable
ax = target_summary["proportion"].plot(
    kind="bar",
    figsize=(7, 4),
    xlabel="Booking Status",
    ylabel="Proportion",
    title="Distribution of Hotel Booking Cancellations",
    rot=0
)

ax.set_xticks([0, 1], labels=["Not Canceled", "Canceled"])
ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 1)

for i, proportion in enumerate(target_summary["proportion"]):
    ax.text(
        i,
        proportion + 0.01,
        f"{proportion:.2%}",
        ha="center"
    )

plt.tight_layout()
plt.show()



# Group lead time and calculate cancellation rates
lead_time_group = pd.cut(
    train_df["lead_time"],
    bins=[-1, 7, 30, 90, 180, 365, np.inf],
    labels=["0–7", "8–30", "31–90", "91–180", "181–365", "366+"]
)

lead_time_summary = (
    train_df.groupby(lead_time_group, observed=False)["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(lead_time_summary)



ax = lead_time_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 4),
    xlabel="Lead Time (Days)",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Lead Time",
    rot=0
)

ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 0.5)


for i, rate in enumerate(lead_time_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.2%}",
        ha="center"
    )



plt.tight_layout()
plt.show()



# Summarize cancellations by deposit type
deposit_summary = (
    train_df.groupby("deposit_type")["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
    .sort_values("cancellation_rate")
)

display(deposit_summary)

# Plot cancellation rate by deposit type
ax = deposit_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 4),
    xlabel="Deposit Type",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Deposit Type",
    rot=0
)


ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 1)

for i, rate in enumerate(deposit_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.2%}",
        ha="center"
    )


plt.tight_layout()
plt.show()

# Group previous cancellations into interpretable categories
previous_cancellation_group = pd.cut(
    train_df["previous_cancellations"],
    bins=[-1, 0, 1, np.inf],
    labels=["0", "1", "2+"]
)

previous_cancellation_summary = (
    train_df.groupby(
        previous_cancellation_group,
        observed=False
    )["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(previous_cancellation_summary)

# Plot cancellation rate by previous cancellation history
ax = previous_cancellation_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 4),
    xlabel="Number of Previous Cancellations",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Previous Cancellation History",
    rot=0
)

ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 1)

for i, rate in enumerate(previous_cancellation_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.2%}",
        ha="center"
    )



plt.tight_layout()
plt.show()

# Group previous non-canceled bookings into interpretable categories
previous_booking_group = pd.cut(
    train_df["previous_bookings_not_canceled"],
    bins=[-1, 0, 1, 5, np.inf],
    labels=["0", "1", "2–5", "6+"]
)

previous_booking_summary = (
    train_df.groupby(
        previous_booking_group,
        observed=False
    )["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(previous_booking_summary)

# Plot cancellation rate by previous booking history
ax = previous_booking_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 4),
    xlabel="Previous Non-Canceled Bookings",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Previous Booking History",
    rot=0
)

ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 0.4)
for i, rate in enumerate(previous_booking_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.2%}",
        ha="center"
    )


plt.tight_layout()
plt.show()

# Group the number of special requests into interpretable categories
special_request_group = pd.cut(
    train_df["total_of_special_requests"],
    bins=[-1, 0, 1, 2, np.inf],
    labels=["0", "1", "2", "3+"]
)

special_request_summary = (
    train_df.groupby(
        special_request_group,
        observed=False
    )["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(special_request_summary)

# Plot cancellation rate by number of special requests
ax = special_request_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 4),
    xlabel="Number of Special Requests",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Number of Special Requests",
    rot=0
)

ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 0.4)

for i, rate in enumerate(special_request_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.2%}",
        ha="center"
    )



plt.tight_layout()
plt.show()


# Group booking changes into interpretable categories
booking_change_group = pd.cut(
    train_df["booking_changes"],
    bins=[-1, 0, 1, 2, np.inf],
    labels=["0", "1", "2", "3+"]
)

booking_change_summary = (
    train_df.groupby(
        booking_change_group,
        observed=False
    )["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(booking_change_summary)

# Plot cancellation rate by number of booking changes
ax = booking_change_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 4),
    xlabel="Number of Booking Changes",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Number of Booking Changes",
    rot=0
)

ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 0.5)

for i, rate in enumerate(booking_change_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.2%}",
        ha="center"
    )

plt.tight_layout()
plt.show()




# Summarize cancellation rates by market segment
market_segment_summary = (
    train_df.groupby("market_segment")["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
    .sort_values("cancellation_rate")
)

display(market_segment_summary)


# Exclude extremely small categories from the visualization only
market_segment_plot = market_segment_summary[
    market_segment_summary["bookings"] >= 100
]


# Plot cancellation rate by market segment
ax = market_segment_plot["cancellation_rate"].plot(
    kind="barh",
    figsize=(8, 5),
    xlabel="Cancellation Rate",
    ylabel="Market Segment",
    title="Cancellation Rate by Market Segment (n>=100)"
)


ax.xaxis.set_major_formatter(PercentFormatter(1))
ax.set_xlim(0, 0.4)

for i, rate in enumerate(market_segment_plot["cancellation_rate"]):
    ax.text(rate + 0.005, i, f"{rate:.2%}", va="center")


plt.tight_layout()
plt.show()



# Audit the original parking-space categories
parking_raw_audit = (
    train_df.groupby("required_car_parking_spaces")["is_canceled"]
    .agg(
        bookings="size",
        cancellations="sum",
        cancellation_rate="mean"
    )
)

display(parking_raw_audit)

# More than 99% of positive values are 1, so pool sparse
# positive categories into a single binary indicator
has_parking_request = train_df["required_car_parking_spaces"] > 0

parking_summary = (
    train_df.groupby(has_parking_request)["is_canceled"]
    .agg(bookings="size", cancellation_rate="mean")
)

display(parking_summary)

# Plot cancellation rate by parking-space requirement
ax = parking_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 4),
    xlabel="Requires Parking Space",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Parking-Space Requirement",
    rot=0
)
ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 0.4)

for i, rate in enumerate(parking_summary["cancellation_rate"]):
    ax.text(i, rate + 0.01, f"{rate:.2%}", ha="center")

plt.tight_layout()
plt.show()


# Check whether the complete-separation pattern holds across
# hotel type and arrival year
parking_hotel_audit = (
    train_df.assign(has_parking_request=has_parking_request)
    .groupby(["hotel", "has_parking_request"])["is_canceled"]
    .agg(bookings="size", cancellations="sum", cancellation_rate="mean")
)

parking_year_audit = (
    train_df.assign(has_parking_request=has_parking_request)
    .groupby(["arrival_date_year", "has_parking_request"])["is_canceled"]
    .agg(bookings="size", cancellations="sum", cancellation_rate="mean")
)

# Summarize the zero-cancellation pattern across hotels and years
parking_stability = pd.concat(
    {
        "Hotel": parking_hotel_audit.xs(True, level="has_parking_request"),
        "Arrival Year": parking_year_audit.xs(True, level="has_parking_request")
    },
    names=["dimension", "group"]
)

display(parking_stability)



# Compare adr_adj distributions by cancellation status
adr_summary = train_df.groupby("is_canceled")["adr_adj"].describe()
display(adr_summary)

# Visualize the distribution

ax = train_df.boxplot(
    column="adr_adj",
    by="is_canceled",
    figsize=(7, 4),
    showfliers=False # Hide fliers
)   



plt.suptitle("")
ax.set_title("Adjusted ADR by Cancellation Status")
ax.set_xlabel("Is Canceled (0 = No, 1 = Yes)")
ax.set_ylabel("Adjusted Average Daily Rate")


plt.tight_layout()
plt.show()


# Summarize cancellation rates by hotel type
hotel_summary = (
    train_df.groupby("hotel")["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(hotel_summary)

# Plot cancellation rate by hotel type
ax = hotel_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 4),
    xlabel="Hotel Type",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Hotel Type",
    rot=0
)

ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 0.4)

for i, rate in enumerate(hotel_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.2%}",
        ha="center"
    )

plt.tight_layout()
plt.show()

# Summarize cancellation rates by customer type
customer_type_summary = (
    train_df.groupby("customer_type")["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
    .sort_values("cancellation_rate")
)

display(customer_type_summary)

# Plot cancellation rate by customer type
ax = customer_type_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 4),
    xlabel="Customer Type",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Customer Type",
    rot=0
)

ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 0.4)

for i, rate in enumerate(customer_type_summary["cancellation_rate"]):
    ax.text(
        i,
        rate + 0.01,
        f"{rate:.2%}",
        ha="center"
    )

plt.tight_layout()
plt.show()

# Group lead time using the same intervals as the bivariate analysis
lead_time_group = pd.cut(
    train_df["lead_time"],
    bins=[-1, 7, 30, 90, 180, 365, np.inf],
    labels=["0–7", "8–30", "31–90", "91–180", "181–365", "366+"]
)

# Summarize cancellations within each lead-time interval and hotel
lead_hotel_summary = (
    train_df.groupby(
        [lead_time_group, "hotel"],
        observed=False
    )["is_canceled"]
    .agg(
        bookings="size",
        cancellation_rate="mean"
    )
)

display(lead_hotel_summary)

# Reshape cancellation rates so each hotel becomes a separate column
lead_hotel_rates = (
    lead_hotel_summary["cancellation_rate"]
    .unstack("hotel")
)

# Compare the two hotels across lead-time intervals
ax = lead_hotel_rates.plot(
    kind="line",
    marker="o",
    figsize=(7, 4),
    xlabel="Lead Time (Days, Grouped)",
    ylabel="Cancellation Rate",
    title="Cancellation Rate by Lead Time and Hotel Type",
    rot=0
)

ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set_ylim(0, 1)
ax.legend(title="Hotel Type")

plt.tight_layout()
plt.show()
```
